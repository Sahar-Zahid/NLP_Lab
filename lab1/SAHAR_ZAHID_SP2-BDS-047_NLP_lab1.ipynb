```python
# Task 7 (part a)

text = "NLP is the future of AI research"

print("Starts with NLP:", text.startswith("NLP"))
print("Ends with research:", text.endswith("research"))
print("Position of future:", text.find("future"))


# Task 7 (part b)

tokens = ["hello", "123", "world", "456abc", "python", "789", "NLP3"]

for token in tokens:
    if token.isalpha():
        print(token, "is alphabetic")
    elif token.isdigit():
        print(token, "is numeric")
    else:
        print(token, "is mixed")


# Task 7 (part c)

words = ["running", "jumped", "NLP", "ai2025", "data", "42"]

pure_words = []
pure_numbers = []
mixed_tokens = []

for word in words:
    if word.isalpha():
        pure_words.append(word)
    elif word.isdigit():
        pure_numbers.append(word)
    else:
        mixed_tokens.append(word)

print("Pure words:", pure_words)
print("Pure numbers:", pure_numbers)
print("Mixed tokens:", mixed_tokens)


# Task 7 (part d)

def validate_token(token):
    if token.isalpha():
        return "WORD"
    elif token.isdigit():
        return "NUMBER"
    else:
        return "MIXED"


tokens = ["chat", "99", "gpt4", "bert", "2024", "nlp!"]

for token in tokens:
    print(token, "→", validate_token(token))
```

##### Task 8

```python
# Task 8 (part a)

def clean(text):
    return text.strip().lower()


clean_lambda = lambda text: text.strip().lower()

text = "  HELLO Natural Language Processing  "

print("Using function:", clean(text))
print("Using lambda:", clean_lambda(text))


# Task 8 (part b)

def tokenize(text):
    return text.split()


def remove_short(tokens, n):
    result = []

    for word in tokens:
        if len(word) > n:
            result.append(word)

    return result


text = "I love working with NLP tools in Python daily"

tokens = tokenize(text)
print("Tokens:", tokens)

long_tokens = remove_short(tokens, 3)
print("After removing short words:", long_tokens)


# Task 8 (part c)

def word_lengths(tokens):
    result = {}

    for word in tokens:
        result[word] = len(word)

    return result


tokens = ["language", "model", "ai", "processing", "deep"]

print("Word lengths:", word_lengths(tokens))


# Task 8 (part d)

def nlp_pipeline(text, min_len):
    text = clean(text)
    tokens = tokenize(text)
    tokens = remove_short(tokens, min_len)
    return word_lengths(tokens)


print(nlp_pipeline("  Natural Language Processing is useful  ", 3))
print(nlp_pipeline("  Python is great for NLP  ", 2))
```

##### Task 9

```python
# Task 9 (part a)

words = ["Python", "NLP", "is", "a", "powerful", "tool", "for", "text"]

lowercase_words = list(map(lambda word: word.lower(), words))
print("Lowercase words:", lowercase_words)

word_lengths = list(map(lambda word: len(word), words))
print("Word lengths:", word_lengths)


# Task 9 (part b)

long_words = list(filter(lambda word: len(word) > 3, words))
print("Words longer than 3:", long_words)

t_words = list(filter(lambda word: "t" in word.lower(), words))
print("Words containing 't':", t_words)


# Task 9 (part c)

alphabetical = sorted(words)
print("Alphabetical:", alphabetical)

short_to_long = sorted(words, key=len)
print("Length ascending:", short_to_long)

long_to_short = sorted(words, key=len, reverse=True)
print("Length descending:", long_to_short)


# Task 9 (part d)

from collections import Counter

sentence = "nlp nlp python ai nlp python text ai nlp python text"

words = sentence.split()

counter = Counter(words)

print("Word frequencies:", counter)
print("Top 3:", counter.most_common(3))
print("Python frequency:", counter["python"])
print("Total words:", len(words))
print("Unique vocabulary:", len(counter))
```
