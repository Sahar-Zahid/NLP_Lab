Task 1

```python
import re
text = "Hi my name is Sahar Zahid. I want to buy these 3 books of 3400 ruoees. Today is 29th of sep 2026"

# part(a)
digits = re.findall(r'\d', text)
print("Digits: ", digits)

# part(b)
capital_words = re.findall(r'\b[A-Z][a-z]*\b', text)
print("Capiitalized words: ", capital_words)

# part(c)
four_digits = re.findall(r'\b\d{4}\b', text)
print("Four_digits: ", four_digits)

# part(d)
range_words = re.findall(r'\b[A-Za-z]{3,6}\b', text)
print("Between 3-6 characters: ", range_words)
```

Task 2

```python
import re

text = """
Order ID ORD-45821 was placed by Sahar at sahar123@gmail.com.
The customer can be contacted at 0300-1234567.
The order was placed on 25/09/2026.
The discount applied was 12.5%, while another customer received 45%.
Please contact support@example.com for further information.
"""

# part(a)
email_pattern = r'[A-za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}'

email = re.findall(email_pattern, text)

print("Emails: ", email)

# part(b)
phone_number = r'03\d{2}-\d{7}'
phone = re.findall(phone_number, text)
print("Phone Number: ", phone)

# part(c)
date_pattern = r'\d{2}/\d{2}/\d{4}'
date = re.findall(date_pattern, text)
print("dates: ", date)

# part(d)
percentage_pattern = r'\d+\.?\d*%'
percentages = re.findall(percentage_pattern, text)

print("Percentages:", percentages)
```

Task 3

```python
paragraph = """
I don't like wasting time on difficult tasks.
It's important to practice NLP regularly because small steps can make learning easier.
My friend said, "You can't understand everything in one day."
So, we decided to practice together and improve our skills.
"""

# part(a)
import nltk
nltk.download('punkt')
nltk.download('punkt_tab')
nltk_tokens = nltk.word_tokenize(paragraph)

print("NLTK Tokens:")
print(nltk_tokens)

# part(b)
import spacy
nlp = spacy.blank("en")

doc = nlp(paragraph)

spacy_tokens = [token.text for token in doc]

print("\nspaCy Tokens:")
print(spacy_tokens)

# part(c)
# Difference 1: NLTK and spaCy use different tokenization rules.
# Difference 2: Contractions such as "don't", "it's", and "can't" are split into separate tokens.

# part(d)
print("\nTotal NLTK Tokens:", len(nltk_tokens))
print("Total spaCy Tokens:", len(spacy_tokens))
```

Task 4

```python
import spacy

nlp = spacy.blank("en")

sentence = "Shantul paid $50 for 3 books. Her email is shantul@example.com."

# part(a)
doc = nlp(sentence)
print("Sentence loaded into spaCy Doc successfully.")

# part(b)
print("\nis_alpha:")
for token in doc:
    print(token.text, ":", token.is_alpha)

# part(c)
print("\nlike_num and like_email:")
for token in doc:
    print(token.text, "like_num:", token.like_num, "like_email:", token.like_email)

# part(d)
print("\nCurrency Tokens:")
for token in doc:
    if token.is_currency:
        print(token.text)
```

Task 5

```python
import spacy

# part(a)
nlp = spacy.load("en_core_web_sm")

paragraph = """
I visited the hospital yesterday. Dr. Ahmed explained the treatment carefully,
and he answered all my questions. The appointment ended at 4 PM.
"""

doc = nlp(paragraph)

# part(b)
print("Sentences:")
for sentence in doc.sents:
    print(sentence.text.strip())

# part(c)
sentences = list(doc.sents)
print("\nTotal Number of Sentences:", len(sentences))

# part(d)
# The abbreviation "Dr." did not cause an incorrect sentence break spaCy correctly kept "Dr. Ahmed" in the same sentence.
```
