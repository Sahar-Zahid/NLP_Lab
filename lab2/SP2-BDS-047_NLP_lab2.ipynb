```python
# part 1
import re
from PyPDF2 import PdfReader

pdf_path = "sample.pdf"

reader = PdfReader(pdf_path)

pages = []

for page in reader.pages:
    text = page.extract_text()
    pages.append(text if text else "")

print("Total Pages:", len(pages))
```

```python
# part 2
all_text = " ".join(pages)

clean_text = re.sub(r'[^\w\s]', '', all_text)
clean_text = re.sub(r'\d+', '', clean_text)

tokens = clean_text.lower().split()

print("Total Tokens:", len(tokens))
```

```python
# part 3
from collections import Counter

stopwords = {
    "the", "is", "a", "an", "and", "or",
    "of", "to", "in", "on", "for",
    "with", "this", "that", "it",
    "as", "was", "are", "be", "by"
}

filtered_tokens = [
    word for word in tokens
    if word not in stopwords
]

frequency = Counter(filtered_tokens)

print("Top 10 Words:")

for word, count in frequency.most_common(10):
    print(word, ":", count)
```

```python
# part 4
keyword = input("Enter keyword: ")

found = False

for page_number, page_text in enumerate(pages, start=1):

    sentences = re.split(r'[.!?]', page_text)

    for sentence in sentences:

        if keyword.lower() in sentence.lower():
            print("Page", page_number, ":", sentence.strip())
            found = True

if not found:
    print("Keyword not found.")
```

#### Task 2

```python
# part 1
from PyPDF2 import PdfReader

pdf_path = "sample.pdf"

reader = PdfReader(pdf_path)

pages_data = []

for i, page in enumerate(reader.pages, start=1):

    raw_text = page.extract_text() or ""

    words = raw_text.split()

    char_count = len(raw_text.replace(" ", ""))

    page_data = {
        "page_number": i,
        "raw_text": raw_text,
        "word_count": len(words),
        "char_count": char_count
    }

    pages_data.append(page_data)

print(pages_data[:2])
```

```python
# part 2
highest = max(pages_data, key=lambda x: x["word_count"])

non_zero_pages = [
    page for page in pages_data
    if page["word_count"] > 0
]

lowest = min(non_zero_pages, key=lambda x: x["word_count"])

average = sum(
    page["word_count"] for page in pages_data
) / len(pages_data)

print("Highest Word Count Page:")
print(highest)

print("\nLowest Non-Zero Word Count Page:")
print(lowest)

print("\nAverage Word Count:", round(average, 2))
```

```python
# part 3
class PhraseNotFoundError(Exception):
    pass


phrase = input("Enter search phrase: ")

matches = []

for page in pages_data:

    count = page["raw_text"].lower().count(
        phrase.lower()
    )

    if count > 0:
        matches.append(
            (page["page_number"], count)
        )

try:

    if not matches:
        raise PhraseNotFoundError(
            "Phrase was not found in any page."
        )

    for page_number, count in matches:
        print(
            "Page:", page_number,
            "Occurrences:", count
        )

except PhraseNotFoundError as e:
    print(e)
```

```python
# part 4
import csv

with open("pdf_summary.csv", "w", newline="") as file:

    writer = csv.writer(file)

    writer.writerow([
        "page_number",
        "word_count",
        "char_count"
    ])

    for page in pages_data:

        writer.writerow([
            page["page_number"],
            page["word_count"],
            page["char_count"]
        ])

print("CSV file created.")

with open("pdf_summary.csv", "r") as file:

    reader = csv.reader(file)

    for row in reader:
        print(row)
```

```python
# part 1
import re
from PyPDF2 import PdfReader

reader = PdfReader("sample.pdf")

text = ""

for page in reader.pages:
    text += page.extract_text() or ""

before = len(text)

text = re.sub(r'https?://\S+|www\.\S+', '', text)
text = re.sub(r'\d+', '', text)
text = re.sub(r'[^A-Za-z.\s]', '', text)
text = re.sub(r'\s+', ' ', text)

after = len(text)

print("Characters Before:", before)
print("Characters After:", after)
```

```python
# part 2
text = text.lower()

sentences = re.split(r'[.!?]', text)

print("Sentence Count Before:", len(sentences))

sentences = [
    sentence.strip()
    for sentence in sentences
    if sentence.strip()
]

print("Sentence Count After:", len(sentences))
```

```python
# part 3
import nltk

nltk.download("stopwords")

from nltk.corpus import stopwords

stop_words = set(
    stopwords.words("english")
)

filtered_sentences = []

for sentence in sentences:

    tokens = sentence.split()

    tokens = [
        token
        for token in tokens
        if token not in stop_words
    ]

    if len(tokens) >= 5:
        filtered_sentences.append(
            " ".join(tokens)
        )

print("Before Filtering:", len(sentences))
print("After Filtering:", len(filtered_sentences))

print("\nThree Sample Sentences:")

for sentence in filtered_sentences[:3]:
    print(sentence)
```

```python
# part 4
with open("nlp_ready_output.txt", "w") as file:

    for sentence in filtered_sentences:
        file.write(sentence + "\n")

with open("nlp_ready_output.txt", "r") as file:

    lines = [
        line.strip()
        for line in file
        if line.strip()
    ]

print("Total Lines:", len(lines))

longest = max(lines, key=len)
shortest = min(lines, key=len)

print("\nLongest Sentence:")
print(longest)

print("\nShortest Sentence:")
print(shortest)
```

```python
# part 1
from docx import Document

doc = Document("sample.docx")

paragraphs = []

for paragraph in doc.paragraphs:

    paragraphs.append({
        "text": paragraph.text,
        "style": paragraph.style.name
    })

print("Total Paragraphs:", len(paragraphs))

styles = set(
    paragraph["style"]
    for paragraph in paragraphs
)

print("\nUnique Styles:")

for style in styles:
    print(style)
```

```python
# part 2
headings = []
body = []
other = []

for paragraph in paragraphs:

    text = paragraph["text"]
    style = paragraph["style"]

    if style.startswith("Heading"):
        headings.append(paragraph)

    elif style == "Normal" or "Body" in style:
        body.append(paragraph)

    else:
        other.append(paragraph)

print("Headings:", len(headings))
print("Body:", len(body))
print("Other/Empty:", len(other))
```

```python
# part 3
qualifying = []

for paragraph in body:

    word_count = len(paragraph["text"].split())

    if word_count > 5:

        char_count = len(
            paragraph["text"].replace(" ", "")
        )

        qualifying.append({
            "text": paragraph["text"],
            "word_count": word_count,
            "char_count": char_count
        })

highest = max(
    qualifying,
    key=lambda x: x["word_count"]
)

lowest = min(
    qualifying,
    key=lambda x: x["word_count"]
)

mean = sum(
    p["word_count"] for p in qualifying
) / len(qualifying)

print("Highest:", highest)
print("Lowest:", lowest)
print("Mean:", round(mean, 2))
```

```python
# part 4
from collections import defaultdict

style_data = defaultdict(
    lambda: {
        "count": 0,
        "texts": []
    }
)

for paragraph in paragraphs:

    style = paragraph["style"]

    style_data[style]["count"] += 1

    if len(style_data[style]["texts"]) < 2:
        style_data[style]["texts"].append(
            paragraph["text"]
        )

sorted_styles = dict(
    sorted(
        style_data.items(),
        key=lambda x: x[1]["count"],
        reverse=True
    )
)

from pprint import pprint

pprint(dict(sorted_styles))
```

```python
from docx import Document

doc = Document("sample.docx")

tables_data = []

for table in doc.tables:

    table_data = []

    for row in table.rows:

        row_data = [
            cell.text
            for cell in row.cells
        ]

        table_data.append(row_data)

    tables_data.append(table_data)

print("Total Tables:", len(tables_data))

for table in tables_data:

    rows = len(table)
    columns = len(table[0])

    print(
        "Dimensions:",
        rows,
        "x",
        columns
    )
```

```python
import pandas as pd

first_table = tables_data[0]

df = pd.DataFrame(
    first_table[1:],
    columns=first_table[0]
)

print(df)

print("\nEmpty Cells Per Column:")
print(df.isna().sum())
```

```python
from docx.oxml.ns import qn

pre_table_paragraphs = []

for element in doc.element.body:

    if element.tag == qn("w:tbl"):
        break

    if element.tag == qn("w:p"):

        text = "".join(
            node.text or ""
            for node in element.iter()
            if node.tag == qn("w:t")
        )

        pre_table_paragraphs.append(text)

print(
    "Pre-table Paragraphs:",
    len(pre_table_paragraphs)
)
```

```python
with open("docx_extracted.txt", "w") as file:

    file.write("--- PRE-TABLE PARAGRAPHS ---\n")

    for paragraph in pre_table_paragraphs:
        file.write(paragraph + "\n")

    file.write("\n--- TABLE DATA ---\n")

    for row in first_table:

        file.write(
            " | ".join(row) + "\n"
        )

print("File created successfully.")
```

```python
from docx import Document
import re

doc_a = Document("sample.docx")
doc_b = Document("Lab 02 Task.docx")

headings_a = [
    p.text.strip()
    for p in doc_a.paragraphs
    if p.style.name.startswith("Heading")
]

headings_b = [
    p.text.strip()
    for p in doc_b.paragraphs
    if p.style.name.startswith("Heading")
]

set_a = set(headings_a)
set_b = set(headings_b)

print("Common:", set_a & set_b)
print("Only A:", set_a - set_b)
print("Only B:", set_b - set_a)
```

```python
normal_a = [
    p.text.strip().lower()
    for p in doc_a.paragraphs
    if p.style.name == "Normal"
]

normal_b = [
    p.text.strip().lower()
    for p in doc_b.paragraphs
    if p.style.name == "Normal"
]

combined = normal_a + normal_b

print("Before Deduplication:", len(combined))

deduplicated = list(set(combined))

print("After Deduplication:", len(deduplicated))
```

```python
from collections import Counter

all_words = []

for paragraph in deduplicated:

    paragraph = re.sub(
        r'[^A-Za-z\s]',
        '',
        paragraph
    )

    words = paragraph.lower().split()

    all_words.extend(words)

stopwords = {
    "the", "is", "a", "an", "and",
    "or", "of", "to", "in", "on",
    "for", "with", "this", "that",
    "it"
}

words = [
    word for word in all_words
    if word not in stopwords
]

frequency = Counter(words)

print("Top 20 Words:")

for word, count in frequency.most_common(20):
    print(word, count)
```

```python
vocabulary_size = len(set(words))

average_length = sum(
    len(p.split())
    for p in deduplicated
) / len(deduplicated)

longest = max(
    deduplicated,
    key=lambda p: len(p.split())
)

print("Total Headings:", len(set_a) + len(set_b))
print("Unique Body Paragraphs:", len(deduplicated))
print("Vocabulary Size:", vocabulary_size)
print("Average Paragraph Length:", round(average_length, 2))
print("Longest Paragraph:", longest)
```

```python
import requests

url = "https://jsonplaceholder.typicode.com/todos"

response = requests.get(url)

if response.status_code != 200:
    print("Error:", response.status_code)
else:

    todos = response.json()

    if not isinstance(todos, list):
        print("Error: Response is not a list.")

    elif len(todos) == 0:
        print("Error: No records found.")

    else:
        print(
            "Total Todo Records:",
            len(todos)
        )
```

```python
from collections import defaultdict

user_todos = defaultdict(
    lambda: {
        "total": 0,
        "completed": 0,
        "incomplete": 0
    }
)

for todo in todos:

    user_id = todo["userId"]

    user_todos[user_id]["total"] += 1

    if todo["completed"]:
        user_todos[user_id]["completed"] += 1
    else:
        user_todos[user_id]["incomplete"] += 1

for user_id, data in user_todos.items():

    valid = (
        data["completed"] +
        data["incomplete"]
        == data["total"]
    )

    print(user_id, data, "Valid:", valid)
```

```python
rates = {}

for user_id, data in user_todos.items():

    rate = (
        data["completed"]
        / data["total"]
    ) * 100

    rates[user_id] = round(rate, 1)

highest = max(rates, key=rates.get)
lowest = min(rates, key=rates.get)

average = sum(rates.values()) / len(rates)

print(
    "Highest:",
    highest,
    rates[highest], "%"
)

print(
    "Lowest:",
    lowest,
    rates[lowest], "%"
)

print(
    "Average:",
    round(average, 1),
    "%"
)
```

```python
users_response = requests.get(
    "https://jsonplaceholder.typicode.com/users"
)

users = users_response.json()

names = {
    user["id"]: user["name"]
    for user in users
}

ranking = sorted(
    user_todos.items(),
    key=lambda x: rates[x[0]],
    reverse=True
)

print(
    f"{'Rank':<6}"
    f"{'Name':<25}"
    f"{'Total':<10}"
    f"{'Completed':<12}"
    f"{'Rate %':<10}"
)

for rank, (user_id, data) in enumerate(
    ranking,
    start=1
):

    print(
        f"{rank:<6}"
        f"{names[user_id]:<25}"
        f"{data['total']:<10}"
        f"{data['completed']:<12}"
        f"{rates[user_id]:<10}"
    )
```

```python
import requests

users = requests.get(
    "https://jsonplaceholder.typicode.com/users"
).json()


def flatten_user(user):

    return {
        "id": user["id"],
        "name": user["name"],
        "username": user["username"],
        "email": user["email"],
        "city": user["address"]["city"],
        "zipcode": user["address"]["zipcode"],
        "lat": user["address"]["geo"]["lat"],
        "lng": user["address"]["geo"]["lng"],
        "company_name": user["company"]["name"],
        "company_catchphrase": user["company"]["catchPhrase"]
    }


flattened_users = [
    flatten_user(user)
    for user in users
]

print(flattened_users)
```

```python
import pandas as pd
import math

df = pd.DataFrame(flattened_users)

df["lat"] = df["lat"].astype(float)
df["lng"] = df["lng"].astype(float)

df["distance_from_origin"] = df.apply(
    lambda row: math.sqrt(
        row["lat"] ** 2 +
        row["lng"] ** 2
    ),
    axis=1
)

print(df)

farthest = df.loc[
    df["distance_from_origin"].idxmax()
]

print("\nFarthest User:")
print(farthest)
```

```python
northern = df[df["lat"] > 0]
southern = df[df["lat"] < 0]

print("Northern Hemisphere:")
print(northern["name"].tolist())

print("\nSouthern Hemisphere:")
print(southern["name"].tolist())

north_average = northern[
    "distance_from_origin"
].mean()

south_average = southern[
    "distance_from_origin"
].mean()

print("\nNorthern Average:", north_average)
print("Southern Average:", south_average)
```

```python
posts = requests.get(
    "https://jsonplaceholder.typicode.com/posts"
).json()

comments = requests.get(
    "https://jsonplaceholder.typicode.com/comments"
).json()

comment_counts = []

for post in posts:

    count = sum(
        comment["postId"] == post["id"]
        for comment in comments
    )

    comment_counts.append(
        (
            post["id"],
            post["title"][:40],
            count
        )
    )

comment_counts.sort(
    key=lambda x: x[2],
    reverse=True
)

print("Top 5 Most-Commented:")

for item in comment_counts[:5]:
    print(item)

print("\nBottom 5 Least-Commented:")

for item in comment_counts[-5:]:
    print(item)
```
